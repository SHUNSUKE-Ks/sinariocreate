````javascript
/*
Runnable demo (no JSX) — uses React / ReactDOM from CDN and renders a simple dialog box.
Paste into a JavaScript-capable Jupyter kernel (ijavascript) and run the cell.
*/

(function(){
  // load React and ReactDOM if not present
  function loadScript(src, cb){
    const s = document.createElement('script'); s.src = src; s.onload = cb; document.head.appendChild(s);
  }
  const needReact = (typeof React === 'undefined');
  const rootId = 'novel-demo-root';
  if(!document.getElementById(rootId)){
    const c = document.createElement('div'); c.id = rootId; document.body.appendChild(c);
  }
  function start(){
    const e = React.createElement;

    function Dialogue({lines}){
      const [index, setIndex] = React.useState(0);
      const cur = lines[index];
      const next = ()=> setIndex(i=> Math.min(i+1, lines.length-1));
      const prev = ()=> setIndex(i=> Math.max(i-1, 0));
      return e('div', {style:{fontFamily:'Yu Gothic, system-ui, sans-serif', padding:16, maxWidth:520, background:'#0f1724', color:'#e6eef8', borderRadius:12, boxShadow:'0 6px 18px rgba(0,0,0,0.4)'}}, [
        e('h3', {key:'h3', style:{margin:'0 0 8px 0', fontSize:18}}, 'ノベル会話デモ'),
        e('div', {key:'meta', style:{opacity:0.9, fontSize:12, marginBottom:8}}, 'tags: ' + (cur.tags||[]).join(', ')),
        e('div', {key:'box', style:{padding:12, background:'#071029', borderRadius:8, minHeight:80}}, [
          e('div', {key:'sp', style:{fontWeight:700, marginBottom:6}}, cur.speaker),
          e('div', {key:'txt', style:{whiteSpace:'pre-wrap', lineHeight:1.6}}, cur.text)
        ]),
        e('div', {key:'controls', style:{marginTop:12, display:'flex', gap:8}}, [
          e('button', {key:'prev', onClick:prev, disabled:index===0, style:{padding:'8px 12px', borderRadius:8, border:'none', cursor:'pointer'}}, '◀︎ 前へ'),
          e('button', {key:'next', onClick:next, disabled=index===lines.length-1, style:{padding:'8px 12px', borderRadius:8, border:'none', cursor:'pointer'}}, '次へ ▶︎'),
          e('div', {key:'pos', style:{marginLeft:'auto', alignSelf:'center', opacity:0.8}}, (index+1) + ' / ' + lines.length)
        ])
      ]);
    }

    const sample = [
      {speaker:'主人公', text:'この剣は...まさか伝説の！', tags:['bg_cave_mysterious','chara_hero_normal','item_sword_legendary','bgm_mystery_forest','se_sword_draw']},
      {speaker:'謎の声', text:'そうだ。長く眠っていたが、お前を選んだ。', tags:['se_echo','voice_mysterious']},
      {speaker:'主人公', text:'選んだ、だと？どういう意味だ……', tags:['chara_hero_confused']},
      {speaker:'ナレーション', text:'風が洞窟の奥から吹き抜けた。剣の刃が淡く光る。', tags:['sfx_wind','effect_glow']}
    ];

    ReactDOM.createRoot(document.getElementById(rootId)).render(
      e(Dialogue, {lines: sample})
    );
  }

  if(needReact){
    loadScript('https://unpkg.com/react@18/umd/react.development.js', ()=>{
      loadScript('https://unpkg.com/react-dom@18/umd/react-dom.development.js', start);
    });
  } else {
    start();
  }
})();
````

## JSX 版（Vite / React 19 プロジェクトにそのまま貼れる）

以下は `src/GameCollections/NovelDialogue.jsx` として貼れる **JSX** の例です。

```jsx
import React, { useState } from 'react';

export default function NovelDialogue({lines = []}){
  const [index, setIndex] = useState(0);
  const cur = lines[index] || {speaker:'', text:'', tags:[]};
  return (
    <div className="novel-dialogue">
      <h3>ノベル会話デモ</h3>
      <div className="tags">{(cur.tags||[]).join(', ')}</div>
      <div className="textbox">
        <div className="speaker">{cur.speaker}</div>
        <div className="text">{cur.text}</div>
      </div>
      <div className="controls">
        <button onClick={()=>setIndex(i=>Math.max(i-1,0))} disabled={index===0}>◀︎ 前へ</button>
        <button onClick={()=>setIndex(i=>Math.min(i+1, lines.length-1))} disabled={index===lines.length-1}>次へ ▶︎</button>
        <div className="pos">{index+1} / {lines.length}</div>
      </div>
    </div>
  );
}
```

```

### コンポーネント説明（一言）
- `NovelDialogue`：会話の表示と前後移動を担当。`lines` に配列を渡すとそれをシーケンス表示します。
- props `lines`：`{speaker, text, tags[]}` の配列（会話データ）。
- CSS の意味：`textbox` はセリフ領域、`controls` は操作ボタンのレイアウト。

    "tags": ["bg_cave_mysterious","chara_hero_normal","item_sword_legendary","bgm_mystery_forest","se_sword_draw"]
  }

```

### Component を使用する際に変更できるプロパティ一覧（props）
- `lines`（必須）: 会話配列（`{speaker,text,tags}`）
- `initialIndex`（任意）: 最初に表示するインデックス
- `onAdvance(index)`（任意）: 次へ/前へ 時に呼ばれるコールバック
- `className`（任意）: 親クラス名でスタイルを上書き可能
